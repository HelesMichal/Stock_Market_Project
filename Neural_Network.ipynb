```python
import pandas as pd
from pathlib import Path
```

```python
frames = []
for file in Path("archive/Stocks").glob("*.txt"):
# for file in sorted(Path("archive/Stocks").glob("*.txt"))[:100]:

    if file.stat().st_size == 0:   # some files are empty
        continue
    df = pd.read_csv(file, parse_dates=["Date"])
    df["Ticker"] = file.name.split(".")[0].upper()
    frames.append(df)

dataset = pd.concat(frames, ignore_index=True)
dataset.tail()
```
